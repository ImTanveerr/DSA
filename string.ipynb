### Reverse

```typescript
const str ="hello";
let reversed ="";
for (let i=str.length-1; i>=0; i--){
    reversed += str[i];
}
```

### Palindrom check

```typescript
function isPalindrome(str){
    let reversed = str.split("").reverse().join("");
    return str==reversed;
}
console.log(isPalindrome("madam")); // true
console.log(isPalindrome("hello"));
```
