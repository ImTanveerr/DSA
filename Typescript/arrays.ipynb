### Reverse array

```typescript
function reverseAray(arr){
    let result=[];
    for(let i=arr.length -1; i>=0;i--){
        result.push(arr[i]);
    }
    return result;
}
console.log(reverseAray([1,3,5,7,9]))
```

### Find max from an array

```typescript
const arr =[10,5,8,20,23];
let max=arr[0];
for(let num of arr){
    if(num>max){
        max=num;
    }
}
```

### Find 2nd largest

```typescript
function secondLargest(arr){
    let unique =[... new Set( arr)];
    unique.sort((a,b)=> b-a);
    return unique[1];
}
console.log(secondLargest([10,1, 5, 20, 20, 8]));
```

### Even or Odd

```typescript
function checkEvenOdd(num){
    if(num %2 ===0){
        console.log("Even");
    }
    else{
        console.log("Odd")
    }
}
checkEvenOdd(10);
checkEvenOdd(11);
```

### Leap year

```typescript
function isLeapYear(year) {
  if ((year % 4 === 0 && year % 100 !== 0) || year % 400 === 0) {
    console.log(year, "is a Leap Year");
  } else {
    console.log(year, "is NOT a Leap Year");
  }
}

isLeapYear(2024); // Leap Year
isLeapYear(2023); // Not Leap Year
```

## Factorial

```typescript
function factorial(n){
    let result =1;
    for(let i=1;i<=n;i++){
        result *=i;
    }
    return result;
}
console.log(factorial(5));
```

### Prime or not

```typescript
function isPrime(num){
    if(num<2) return false;
    
    for(let i=2;i<=Math.sqrt(num);i++){
        if(num % i=== 0){
            return false;
        }        
        else{
            return true;
        }
    }
}
console.log(isPrime(7));  // true
console.log(isPrime(10)); // false
```
