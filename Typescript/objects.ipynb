### Get all keys of an object

```typescript
const person = { name: "Akram", age: 22, skill: "React" };
console.log(Object.keys(person)); // ["name","age","skill"]
console.log(Object.values(person)); // ["name","age","skill"]
```

### Check if property exists


```typescript
console.log("age" in person);  // true
console.log("salary" in person); // false
```

### Merge two objects

```typescript
const obj1 = { a: 1, b: 2 };
const obj2 = { b: 3, c: 4 };
const merged = { ...obj1, ...obj2 };
console.log(merged); // { a:1, b:3, c:4 }
```
